# Day 1 — Understanding Underwriting Rules

This task focuses on understanding how banks evaluate loan applicants and identify risky customers. The `german_credit_data.csv` dataset is used to analyze factors like income, savings, job stability, credit amount, and loan duration to classify customers as good or bad credit risks.

```python
# Load Dataset
df = pd.read_csv('german_credit_data.csv')

# Handle Missing Values
# 1. Employment length: fill with 0 (assuming unemployed/student)
df['person_emp_length'].fillna(0, inplace=True)
# 2. Interest rate: fill with median (industry standard for missing rates)
df['loan_int_rate'].fillna(df['loan_int_rate'].median(), inplace=True)

# Remove unrealistic outliers (e.g., age > 100)
df = df[df['person_age'] < 100]

print(f"Dataset Shape: {df.shape}")
df.head()
```
